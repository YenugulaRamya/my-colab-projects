```python
from qiskit import QuantumCircuit

# Create circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Apply gates
qc.h(0)        # Hadamard gate
qc.cx(0, 1)    # CNOT gate

# Measurement
qc.measure([0,1], [0,1])

qc
```

```python
import matplotlib.pyplot as plt

qc.draw(output='mpl')
plt.show()
```

```python
from qiskit_aer import AerSimulator

# Create simulator
simulator = AerSimulator()

# Run circuit
job = simulator.run(qc, shots=1024)

# Get result
result = job.result()
counts = result.get_counts()

print(counts)
```

```python
from qiskit.visualization import plot_histogram

plot_histogram(counts)
plt.show()
```

```python
from qiskit_aer import StatevectorSimulator
import numpy as np

phi_plus=QuantumCircuit(2)
phi_plus.h(0)
phi_plus.cx(0,1)

# Using StatevectorSimulator to get the statevector
statevector_simulator = StatevectorSimulator()
result=statevector_simulator.run(phi_plus).result()
sv=result.get_statevector()

# Corrected assert statement
assert np.allclose(sv.data,[1/np.sqrt(2),0,0,1/np.sqrt(2)]), "Statevector does not match expected Bell state"
```

```python
from qiskit import QuantumCircuit

# Define phi_minus circuit to produce ( |00⟩ - |11⟩ ) / sqrt(2)
phi_minus = QuantumCircuit(2)
phi_minus.h(0)
phi_minus.cx(0, 1)
phi_minus.z(0)

# Reuse the statevector_simulator from a previous cell (OOkH-Zvw1XRW)
result = statevector_simulator.run(phi_minus).result()
sv = result.get_statevector()

# Assert the statevector
assert np.allclose(sv.data,[1/np.sqrt(2),0,0,-1/np.sqrt(2)]), "Statevector does not match expected phi_minus Bell state"
```

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import StatevectorSimulator # Ensure StatevectorSimulator is imported

psi_plus = QuantumCircuit(2)
psi_plus.x(0) # Apply X gate to the first qubit
psi_plus.h(1) # Apply H gate to the second qubit
psi_plus.cx(1, 0) # Apply CNOT gate with control on qubit 1 and target on qubit 0

# Use StatevectorSimulator to get the statevector
statevector_simulator = StatevectorSimulator()
result = statevector_simulator.run(psi_plus).result()
sv = result.get_statevector()

assert np.allclose(sv.data,[0,1/np.sqrt(2),1/np.sqrt(2),0]), "Statevector does not match expected psi_plus Bell state"
```

```python
from qiskit.quantum_info import SparsePauliOp

# Make a quantum circuit
qc = QuantumCircuit(3)
qc.h(0)
qc.cx(0,1)
qc.cx(0,2)

observable = SparsePauliOp("Z" * 3)
print(f"Observable: {observable.paulis}")
pm = generate_preset_pass_manager(optimization_level=1, backend=backend)
isa_circuit = pm.run(qc)
isa_observable = observable.apply_layout(isa_circuit.layout)
```

```python
estimator = EstimatorV2(mode=backend)
job = estimator.run([(isa_circuit, isa_observable)])
print(job.job_id())
result = job.result()
pub_result = result[0]
print(f"Expectation value: {pub_result.data.evs}")
```

```python
def n_ghz(num_qubits):
  n_ghz=QuantumCircuit(num_qubits)
  n_ghz.h(0)
  for i in range(1,num_qubits):
    n_ghz.cx(0,i)
  return n_ghz
ghz_8=n_ghz(8)
ghz_8.measure_all()
ghz_8.draw('mpl')
  #lets inspect our GHZ circuit
ghz=n_ghz(15)
print(f"Circuit depth: {ghz.depth()}")
print(f"Total gate count: {ghz.size()}")
print(f"Number of qubits: {ghz.num_qubits} ")
print(f"Number of clbits: {ghz.num_clbits}")
print(f"Constituent Gates:  {dict(ghz.count_ops())}")
```
